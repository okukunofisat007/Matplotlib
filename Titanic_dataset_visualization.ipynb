# Data Visualization with Titanic_dataset

### Data visualization is the graphical representation of data and information. 
### It is the process of representing data through visual elements like charts, graphs, and maps, making it easier to understand patterns, trends, and insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline  
```

```python
titanic_df = pd.read_csv('data/titanic.csv')

titanic_df.head()
```

```python
titanic_df.info()
```

```python
# Plot Gender count using BARCHART
gender_counts = titanic_df['Sex'].value_counts()

plt.bar( ['male', 'female'], gender_counts, width = 0.3)
plt.title("Titanic Gender count")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# Plot PASSENGER CLASS using BAR CHART
p_class = titanic_df['Pclass'].value_counts()

plt.bar( ["First class", "Second class", "Third class"], p_class)
plt.title("Titanic Pclass count")
plt.xlabel("P_class")
plt.ylabel("Count")
plt.show()
```

```python
# Plot Survival counts using BARCHART

survival_counts = titanic_df.groupby('Sex')['Survived'].value_counts().unstack()   # unstack makes it easier to read

survival_counts.plot(kind='bar', figsize=(8, 6))
plt.title("Titanic Survival count")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
survival_counts = titanic_df.groupby('Sex')['Survived'].value_counts().unstack()
survival_counts
```

```python
# plot Age distribution with HISTOGRAM

plt.figure(figsize=(10, 7))
plt.hist(titanic_df['Age'], bins=20, color='g', edgecolor = 'r')         # Bins is the number of plot. change it to match your preference
plt.title("Titanic Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

```

```python
# Practical Example: plotting the pie chat for survived in titanic


category_names = ['Died ', 'Survived']
sizes = [titanic_df.Survived.value_counts()[0], titanic_df.Survived.value_counts()[1]]

plt.figure(figsize=(10, 7))
plt.pie(sizes, labels=category_names, autopct='%1.1f%%', explode=[0, 0.05], textprops={'fontsize': 20, 'color': 'red'})
plt.show()
```
